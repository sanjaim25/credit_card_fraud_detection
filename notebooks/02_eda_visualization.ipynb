```python
# 02 - EDA & Visualization
# Exploratory data analysis with visualizations for class imbalance,
# feature distributions, and correlations.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import yaml
import sys
import os

# Add src folder to path
sys.path.append("../src")

from eda_utils import (
    plot_class_distribution,
    plot_correlation_heatmap,
    plot_amount_distribution,
    plot_time_distribution
)

# Load config file
with open("../configs/config.yaml", "r") as f:
    config = yaml.safe_load(f)

# Load dataset
df = pd.read_csv(config["paths"]["raw_data"])

# Figure save location
FIGURES = config["paths"]["figures"]

# Create folder if not exists
os.makedirs(FIGURES, exist_ok=True)

print("Dataset Shape:", df.shape)

# -----------------------------
# Class Distribution
# -----------------------------
plot_class_distribution(
    df,
    save_path=os.path.join(FIGURES, "class_distribution.png")
)

# -----------------------------
# Correlation Heatmap
# -----------------------------
plot_correlation_heatmap(
    df,
    save_path=os.path.join(FIGURES, "correlation_heatmap.png")
)

# -----------------------------
# Amount Distribution
# -----------------------------
plot_amount_distribution(
    df,
    save_path=os.path.join(FIGURES, "amount_distribution.png")
)

# -----------------------------
# Time Distribution
# -----------------------------
plot_time_distribution(
    df,
    save_path=os.path.join(FIGURES, "time_distribution.png")
)
```
